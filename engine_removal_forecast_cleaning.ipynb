### Initial Notes:  
1- We have 7 phases (CRUISE, CLIMB, TAKEOFF, ENGINE START, CEOD_CRUISE, CEOD_TAKEOFF, CEOD_CLIMB).  
2- From knowledge of the business domain, the critical phases (most stress) that determine engine life are takeoff, climb, and cruise.  
3- Also from business domain, each takeoff is considered one running cycle that counts into the engine remaining useful life.  
4- Flights are ordered new to old (last flight of the engine is the first on top).

```python
# Keep only the critical flight phases: CRUISE, TAKEOFF, and CLIMB
df = df[df['flight_phase'].isin(['CRUISE', 'TAKEOFF'])]
df['flight_phase'] = df['flight_phase'].cat.remove_unused_categories()
print(f"Rows after filtering: {len(df)}")
print(df['flight_phase'].unique())
```

```python
# create date and time columns from flight_datetime (format: 'YYYY-MM-DD HH:MM:SS')
df['flight_datetime_c'] = pd.to_datetime(df['flight_datetime'], format='%Y-%m-%d %H:%M:%S')
df['date'] = df['flight_datetime_c'].dt.date
df['time'] = df['flight_datetime_c'].dt.time
df.head(30)
```

```python
#sort the data by engine number and flight datetime
df = df.sort_values(by=['esn', 'flight_datetime_c'])
df.head(30)
```

```python
# drop the flight_datetime column
df = df.drop('flight_datetime', axis=1)
```

```python
# for each engine lets assign a flight cycle for each row, flight cycle start with take off and end with cruise
# flight cycle will increase by 1 for each take off from oldest to newest date for each engine
# first split daata by engine into a list of dataframes 
# then from oldest to newsest date for each engine assign a flight cycle for each row that increaes by 1 for each take off
# then combine the dataframes into one dataframe
df['flight_cycle'] = 0
df_list = [df[df['esn'] == i] for i in df['esn'].unique()]
for i in range(len(df_list)):
    df_list[i] = df_list[i].sort_values(by='flight_datetime_c')
    flight_cycle = 0
    for j in range(len(df_list[i])):
        if df_list[i].iloc[j]['flight_phase'] == 'TAKEOFF':
            flight_cycle += 1
        df_list[i].iloc[j, df_list[i].columns.get_loc('flight_cycle')] = flight_cycle
```

```python
# combine the dataframes
df = pd.concat(df_list)
```

```python
df.head(20)
```

```python
# --- RCL with policy-based removal threshold ---
# Fleet operating policy: engines are removed at POLICY_REMOVAL_CYCLE cycles.
#
# expected_max_cycle per engine:
#   - removed engine that came off below policy: keep observed max_cycle
#     (the engine genuinely retired at fewer cycles)
#   - removed engine that came off above policy: cap at policy
#   - active engine: assume policy threshold (the planned removal point)
#
# This unifies the labels across the whole fleet under a single auditable
# threshold, replacing the earlier empirical-prior approach (which depended
# on the small sample of off-wing engines).

REMOVAL_THRESHOLD_DAYS = 21
POLICY_REMOVAL_CYCLE = 5000

df['flight_datetime_c'] = pd.to_datetime(df['flight_datetime_c'])
last_obs   = df.groupby('esn')['flight_datetime_c'].max()
max_cycles = df.groupby('esn')['flight_cycle'].max()
cutoff = df['flight_datetime_c'].max()
days_since_last = (cutoff - last_obs).dt.days
is_removed = days_since_last >= REMOVAL_THRESHOLD_DAYS

# Apply the policy: cap at POLICY_REMOVAL_CYCLE for engines past it,
# otherwise use observed max_cycle for removed engines, or POLICY_REMOVAL_CYCLE for active.
expected_max = {}
for esn in df['esn'].unique():
    if is_removed.get(esn, False):
        expected_max[esn] = float(min(POLICY_REMOVAL_CYCLE, max_cycles[esn]))
    else:
        expected_max[esn] = float(POLICY_REMOVAL_CYCLE)

print(f"Policy removal cycle: {POLICY_REMOVAL_CYCLE}")
print(f"Removed engines: {int(is_removed.sum())} / {len(is_removed)}")
print(f"expected_max distribution:")
print(pd.Series(expected_max).value_counts().sort_index())

# RCL using the policy-aware anchor, clipped at 0.
df['RCL'] = (df['esn'].map(expected_max) - df['flight_cycle']).clip(lower=0)
max_flight_cycle = max_cycles.to_dict()
df['is_engine_active'] = (~df['esn'].map(is_removed)).astype(int)
df['expected_max_cycle'] = df['esn'].map(expected_max)

print(f"\nRCL summary after policy correction:")
print(df['RCL'].describe())
```

## Combine Engine Types: Baseline Normalisation + One-Hot Encoding

The EDA confirmed that `LEAP-1A26` and `LEAP-1A26E1` share essentially the same operating envelope (11/12 feature/phase comparisons within 5% of each other; only `fuel_flw` in TAKEOFF diverges by ~24%, the known E1 fuel-burn improvement).

To merge both types into one training set without confounding type-offset with degradation, we add two preprocessing steps before saving:

1. **Per-engine baseline normalisation:** for each engine, subtract its first-50-flight-cycles mean per phase per feature. The result (suffix `_norm`) is a drift signal centred on each engine's own healthy baseline, so absolute level differences between -1A26 and -1A26E1 disappear.
2. **One-hot encoding of `eng_type`:** lets sequence models retain the type label as a feature without it leaking through the numeric sensor channels.


```python
# Per-engine baseline normalisation (first N flight cycles per phase)
import numpy as np
import pandas as pd

baseline_cycles = 50  # first 50 cycles considered "healthy baseline"
sensor_features = ['egt_probe_average', 'fuel_flw', 'core_spd', 'oil_temp', 'oil_pres', 'pt2_(fan_inlet_tot_pres)']

def add_baseline_normalised(df, features, n_cycles=50):
    df = df.copy()
    for feat in features:
        df[f'{feat}_norm'] = np.nan
    for esn in df['esn'].unique():
        for phase in df['flight_phase'].cat.categories if hasattr(df['flight_phase'], 'cat') else df['flight_phase'].unique():
            mask = (df['esn'] == esn) & (df['flight_phase'] == phase)
            sub = df.loc[mask].sort_values('flight_cycle')
            if sub.empty:
                continue
            baseline = sub[sub['flight_cycle'] <= n_cycles][features].mean()
            for feat in features:
                df.loc[mask, f'{feat}_norm'] = df.loc[mask, feat] - baseline[feat]
    return df

df = add_baseline_normalised(df, sensor_features, n_cycles=baseline_cycles)
print(f"Added {len(sensor_features)} baseline-normalised columns (suffix _norm).")
print(df[[f'{f}_norm' for f in sensor_features]].describe().round(2))
```

```python
# One-hot encode eng_type so the model can keep the type label without it bleeding into sensor channels
eng_type_dummies = pd.get_dummies(df['eng_type'].astype(str), prefix='engtype').astype(int)
df = pd.concat([df, eng_type_dummies], axis=1)
print("Added one-hot columns:", list(eng_type_dummies.columns))
print(df[list(eng_type_dummies.columns)].sum())
```
